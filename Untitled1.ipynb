```python
#@title 🕹️ E-Commerce Dashboard Control Panel { run: "auto" }
selected_department = "Blouses" #@param ["Dresses", "Knits", "Blouses", "Sweaters", "Jackets", "Bottoms"]

import pandas as pd
import plotly.express as px
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer

# 1. Load the dataset into Pandas
df = pd.read_csv('Womens Clothing E-Commerce Reviews.csv', index_col=0)

# 2. Initialize language toolkit and download lexicon assets
nltk.download('vader_lexicon', quiet=True)
sia = SentimentIntensityAnalyzer()

# 3. Drop missing entries in target columns
clean_df = df.dropna(subset=['Review Text', 'Department Name']).copy()

# 4. Calculate Sentiment Score (-1 = Negative, 1 = Positive)
print("Processing customer sentiment scores...")
clean_df['Sentiment_Score'] = clean_df['Review Text'].apply(lambda x: sia.polarity_scores(str(x))['compound'])

# 5. Apply the interactive sidebar parameter filter
filtered_df = clean_df[clean_df['Department Name'] == selected_department]

# 6. Generate the Interactive Scatter Plot
fig = px.scatter(
    filtered_df,
    x='Sentiment_Score',
    y='Rating',
    color='Recommended IND',
    hover_data=['Title'],
    title=f'Product Sentiment vs Rating Analysis: {selected_department} Segment',
    labels={'Sentiment_Score': 'Customer Sentiment (Negative to Positive)', 'Rating': 'Review Star Rating'},
    color_continuous_scale=px.colors.diverging.RdYlGn
)

# Apply a clean dark theme UI
fig.update_layout(template='plotly_dark', width=1000, height=550)
fig.show()

# 7. Print operational data summary for business teams
st_risk = filtered_df[filtered_df['Sentiment_Score'] < -0.3]
print(f"\n⚠️ Alert: Found {len(st_risk)} high-risk negative reviews in the '{selected_department}' department.")
```
